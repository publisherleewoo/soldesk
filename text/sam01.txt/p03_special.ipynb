```python
# unicode : 컴에 글자를 표현하는 방식 중 하나
# Unicode Transfs Format - 8bit 
from sys import maxunicode
from unicodedata import category

print(maxunicode)  # unicode 문자 전체수
print(chr(65)) # 65번째 unicode 문자
print(chr(66)) # 66번째 unicode 문자
print(category("A")) #A의 즐가
 
```

```python
special = []
for i in range(maxunicode):
    # print(chr(i), category(chr(i)))
    if category(chr(i)).startswith("P"):
        special.append(chr(i))
print(special)
```

```python
f = open('./Jan22_1_Text/sam01.txt','r',encoding='utf-8')


wc = {}
for line in f.readlines():
    for s in special:
        line = line.replace(s,"")
    line = line.replace("\n","").split(" ")
    for word in line:
        if word in wc:
            wc[word] +=1
        else:
            wc[word] = 1

# for w,c in wc.items():
#     print(w,c)

f.close()
```
