```python
# 소설 파일 불러다가

# 동사만 원형으로 갯수

f = open('./Jan22_1_text/sherlock.txt', 'r', encoding="utf-8")
data = f.read()
print(data)
```

```python
from nltk.tokenize import word_tokenize
from nltk.stem.wordnet import WordNetLemmatizer
from nltk.corpus import wordnet

wnl = WordNetLemmatizer() # 표제어추출 (동사원형)
 
 
obj = {}

for w in word_tokenize(data):
    if wnl.lemmatize(w, wordnet.VERB) in obj:
        obj[wnl.lemmatize(w, wordnet.VERB)] +=1
    else:
        obj[wnl.lemmatize(w, wordnet.VERB)] = 0
    
print(obj)
```
