```python
import pandas as pd

df = pd.read_csv('./lnps 1.csv', names=['마트','품명','가격','날짜','종류','구'])
df

# 가격 오타인거 삭제
df.sort_values(by="가격",ascending=False)

# 평균값
df = df[df['가격']<3000000]
df

# 종류별 평균가
df.groupby("종류")["가격"].mean()

# 구-> 종류별 평균가
df.groupby(["구","종류"])["가격"].mean()
```
