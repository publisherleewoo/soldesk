```python
import pandas as pd

a = pd.DataFrame()
a["이름"] = ["새우깡", "양파링"]
a["가격"] = [3000, 4000]
 

s = pd.Series(["초코파이", 5000], index=["이름", "가격"])  #이 Serires를  데이터프레임화 시켜서 concat으로 합치기
s = pd.DataFrame([s])

a = pd.concat([a,s])
print(a)

print('----------------------')

s  = {"이름":"포카칩","가격":4000}  #이 dic을  데이터프레임화 시켜서 concat으로 합치기
s =pd.DataFrame([s])
a =  pd.concat([a,s])
print(a)
```
