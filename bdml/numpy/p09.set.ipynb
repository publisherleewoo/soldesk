```python
import numpy as np
```

```python
a = np.random.randint(1,11,[2,3])
print(a)
```

```python
b = np.random.randint(1,11,[2,3])
print(b)
```

교집합

```python
np.intersect1d(a,b) 
```

합집합

```python
np.union1d(a,b)
```

차집합

```python
np.setdiff1d(a,b)
```

a나 b 한쪽에만 있는

```python
np.setxor1d(a,b) 
```

중복제거

```python
np.unique(a)
```
