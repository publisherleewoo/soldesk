```python
# 연습문제
name = np.array(['홍길동','김길동','이길동'])
kor = np.array([100,90,85])
eng = np.array([10,50,20])
mat = np.array([40,60,55])

print(name[np.greater(np.divide(np.add(np.add(kor,eng),mat),len(name)),60)])
print(name[(kor+eng+mat)/len(name)>60])
```
