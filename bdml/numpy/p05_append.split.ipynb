```python
import numpy as np
```

```python
a = np.random.randint(1,100,[3,5])
b = np.random.randint(1,100,[3,5])
print(a)
print(b)
```

```python
a+b  
```

```python
np.concatenate([a,b])
```

```python
np.append(a,b)
```

```python
np.concatenate([a,b],axis=0)   # axis = 0 열방향 , axis = 1  행방향
```

```python
np.split(a,3)  # 2차원행렬   [[a],[b],[c]]를   [[a]],[[b]],[[c]] 로 

```

```python
np.array_split(a,2)  # 2개의 배열로 (알아서)

```
