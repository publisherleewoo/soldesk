```python
import numpy as np
```

행렬의 모든 합

```python
a = np.random.randint(1,10,[2,3])
print(a)
b = np.sum(a)
print(b)
```

행렬의 모든 퍙균

```python
c =np.mean(a)
print(c)
```

분산 : 평균에서 얼마나 퍼졌나
표준편차

```python
d = a - c  # 각 값에서 평균을 뺌(값이랑 평균이랑 얼마나 차이나나)
print(d)
```

```python
print(np.sum(d)) # 다더하면 0임
```

```python
e = d**2   #더하면 0이기때문에 제곱해서 음수를 없애버림
print(e)

f = np.mean(e)    # 이것이 분산값
print(f)    
```

```python
np.var(a)  #  분산값 메서드
```

```python
np.sqrt(f) # 분산 구한걸 루트를 씌운게 표준편차
```

```python
i = np.std(a) #표준편차
print(i)
```

```python
np.max(a,axis=0)  #행의 최대값 배열
```

```python
np.min(a,axis=1)  #열의 최소값 배열
```

```python
print(a)
```

```python
np.argmax(a)  # 최대값 인덱스(0부터 시작해서 제일큰게 몇번째에)
```

```python
np.argmin(a) # 만약에 같은 숫자가 나오면 앞에 인덱스를 띄움
```
