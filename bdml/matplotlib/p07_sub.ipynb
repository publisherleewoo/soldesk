```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import matplotlib.font_manager as fm

font_path = "C:/Windows/Fonts/malgun.ttf"
font_name = fm.FontProperties(fname=font_path).get_name()
plt.rc("font", family=font_name)
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv(
    "../cspf.csv",
    names=["날짜", "노선", "역", "내고타", "안내고타", "내고내리", "안내고내리"],
)

df = df.groupby("노선")[["내고타", "안내고타", "내고내리", "안내고내리"]].mean()

data1 = df["내고타"].to_numpy()
data2 = df["안내고타"].to_numpy()
data3 = df["내고내리"].to_numpy()
data4 = df["안내고내리"].to_numpy()
xLabel = df.index.to_numpy()
x = np.arange(len(xLabel))


plt.figure(figsize=(10, 5))

plt.bar(x - 0.4, data1, width=0.4, label="내고타")
plt.bar(x - 0.4, data3, width=0.4, bottom=data1, label="내고내리")
plt.bar(x, data2, width=0.4, label="안내고타")
plt.bar(x, data4, width=0.4, bottom=data2, label="안내고내리")

plt.title('지하철')
plt.xticks(x,xLabel,rotation='vertical')

plt.legend()
```

```python
pieData = [data1.mean(),data2.mean(),data3.mean(),data4.mean(),]
pieLabel = ["내고타", "안내고타", "내고내리", "안내고내리"]
plt.pie(pieData,labels=pieLabel,autopct="%.1f%%",colors=["red","orange","blue","skyblue"])
plt.title("지하철") 
```
