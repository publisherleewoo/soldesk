```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.font_manager as fm
import pandas as pd

font_path = 'C:/Windows/Fonts/malgun.ttf' 
font_name = fm.FontProperties(fname=font_path).get_name()
plt.rc('font', family=font_name)
plt.rcParams['axes.unicode_minus'] = False
```

```python
df = pd.read_csv('../numpy/seoulDust.csv',names=["날짜","권역","구","미세","초미세","정도"])
df = df[df["구"]=="종로구"]
df= df.sort_values(by="날짜")

 
df['날짜'] = pd.to_datetime(df['날짜']).dt.strftime('%m/%d %H시')
 
d=df['날짜'].to_numpy()
pm10 = df["미세"].to_numpy()
pm25=df['초미세'].to_numpy()
 
_,sub1Conf = plt.subplots()
p1 = sub1Conf.plot(pm10,"r")
sub1Conf.set_xlabel('시간')
sub1Conf.set_ylabel('미세')

sub2Conf = sub1Conf.twinx()
p2 = sub2Conf.plot(pm25,"b")
sub2Conf.set_ylabel('초미세')

sub1Conf.legend(p1+p2,['미세','초미세'])
plt.title('종로구 미세먼지')
plt.xticks(np.arange(len(d)),d)
 
plt

 
```
