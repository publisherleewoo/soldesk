```python
from datetime import date
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import matplotlib.font_manager as fm
import seaborn as sns

font_path = "C:/Windows/Fonts/malgun.ttf"
font_name = fm.FontProperties(fname=font_path).get_name()
plt.rc("font", family=font_name)
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv(
    "../numpy/subway.csv",
    names=["년", "월", "일", "노선", "역", "타", "내린"],
)

df["이용객수"] = df["타"]+df["내린"]


# 노선별 이용객수 (탄+내린) 평균 막대그래프

# sns.barplot(df, x="노선", y="이용객수", palette="rainbow", hue="노선")

# 요일별 이용수(탄+내린) 평균 막대그래프

df['날짜'] = pd.to_datetime(df[['년', '월', '일']].rename(columns={'년': 'year', '월': 'month', '일': 'day'}))
 
weekday_list = ['월', '화', '수', '목', '금', '토', '일']
 
df['요일'] = df['날짜'].dt.weekday.apply(lambda x: weekday_list[x])

sns.barplot(df, x="요일", y="이용객수", palette="rainbow", hue="요일")
```
