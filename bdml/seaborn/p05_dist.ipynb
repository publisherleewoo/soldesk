```python
import pandas as pd
from http.client import HTTPConnection
from json import loads
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

# 한글 폰트 설정 (Windows의 경우 맑은 고딕)
plt.rcParams['font.family'] = 'Malgun Gothic'

# 마이너스 기호(-)가 깨지는 현상 방지
plt.rcParams['axes.unicode_minus'] = False

hc = HTTPConnection("openapi.seoul.go.kr:8088")
hc.request("GET","/575a4655496b636839386f58586542/json/RealtimeCityAir/1/16/")

res = hc.getresponse()
resBody = res.read()
txt = resBody.decode() 
hc.close()

dustData = loads(resBody)  #JSON =>Python 컬렉션
dustDF = pd.DataFrame(dustData["RealtimeCityAir"]["row"])

print(dustDF)

sns.histplot(dustDF,x="PM")


```
