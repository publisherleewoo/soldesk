```python
import nltk
nltk.download("punkt_tab") #특수문자 관련
```

```python
sentence = "8-8. 백그라운드 실행중인거 종료 ps -ef : 실행중인 프로세스들? 우리 가비아 과정은 AI 어플리케이션(APP) 개발자 직군이고 직무는" 

from nltk.tokenize import word_tokenize,sent_tokenize

wt = word_tokenize(sentence) # 단어 나누기(특수문자 따로)
print(wt)

st = sent_tokenize(sentence) # 문장 나누기
print(st)
```
