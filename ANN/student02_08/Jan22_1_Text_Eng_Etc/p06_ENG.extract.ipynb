```python
# extract(추출하기)
from nltk.tokenize import word_tokenize
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer() # 어간추출(동사원형x)하는 객체 생성

words = "play playing plays played"
wt = word_tokenize(words)
for w in wt:
    print(w,ps.stem(w))
print("-----")

words = "variable variety variation"
wt = word_tokenize(words)
for w in wt:
    print(w,ps.stem(w))
```

```python
from nltk.tokenize import word_tokenize
from nltk.stem.wordnet import WordNetLemmatizer
from nltk.corpus import wordnet

wnl = WordNetLemmatizer() # 표제어추출 (동사원형)
 
words = "am are is"
wt = word_tokenize(words)
for w in wt:
    print(w, "->", wnl.lemmatize(w, wordnet.VERB)) 

print("-----")

 
words = "say said"
wt = word_tokenize(words)
for w in wt:
    print(w, "->", wnl.lemmatize(w, wordnet.VERB))
```
