```python
# linear regression
#   주어진 값들을 최대한 만족시킬만한 a/b를 찾아내서
#   -> 회귀방정식( y = ax+b )을 완성시키고
#   -> 그 식 써서 예측하자

# 인공신경망(DNN) : 다차원 + 다층회귀분석




# [1,10],[2,20],[3,30] ->[4,?]   : y = 10x  -> 40
# [1,11],[2,21],[3,31] ->[4,?]   : y = 10x+1  -> 41
# [1,9],[2,20],[3,28.756] ->[4,?]  : y=ax+b -> ?? 

```

```python
import tensorflow as tf
tf.compat.v1.disable_v2_behavior() # tf 2.x기능 off    1버전대 기능 먼저 공부

```

```python

# 1) 학습데이터 확보 + 전처리
xData = [1,2,3]
yData = [10,20,30]
```

```python
# 2) 인공신경망 구성
#   변수        y = ax + b
#       Variable : a/b  : AI가 찾아내야 할 값
#       placeholder x/y : 학습데이터 들어갈 자리

# 첫 값(0으로, 1차원)  => [0,0,0,0,0] 이런느낌
a = tf.Variable(tf.zeros([1], dtype=tf.float64))

# 첫 값(정규분포에 따르는 -1 ~ 2 사이의 랜덤, 1차원, 자료형은 float)
b = tf.Variable(tf.compat.v1.random_uniform([1], -1,2, dtype=tf.float64))

x = tf.compat.v1.placeholder(tf.float64)
y = tf.compat.v1.placeholder(tf.float64)

sik = x * a + b
# 처음시작 
#       x * 0 + 0.12
#       sik을 통해서 [1,?] 일때  [1,0.12]     y를 sik을 통해서 구하면 x=1  -> y = 0.12
#                                             실제 y 값은            x=1  -> y = 10
#                                           -> 10, 0.12 차이가 너무 크다 =>오류가 있다(MSE) -> a,b를 바꿔보자(경사하강법으로)
# 바꿔서
#       x * 1 + 0.3
#       sik을 통해서 구하면 x=1 -> y=1.3
#       실제 y값은          x=1 -> y=10
#       -> sik을 통해 구한값 (1.3), 실제값 (10)차이가 너무 크다 -> a,b를 바꿔보자
# 반복
#       sik을 통해서 구한 값과 실제 y랑 차이가 별로 없을때까지

# MSE + 차원수 줄이고
loss = tf.reduce_mean(tf.square(y - sik)) #차원수 줄이기(MSE)

# 최적의 a/b를 찾아줄 객체
o = tf.compat.v1.train.GradientDescentOptimizer(learning_rate=0.01) # 경사하강법 최적화

# loss를 최소화 하는게 목표
goal = o.minimize(loss) 



# 3) 학습데이터 넣어서 최적의 a/b찾자
s = tf.compat.v1.Session()
s.run(tf.compat.v1.global_variables_initializer()) # Variable들 첫 값 뽑기

for i in range(10000):
    s.run(goal, feed_dict={x:xData,y:yData}) #placeholder  x에 xData넣고, y에 yData넣고
    # print(s.run(loss, feed_dict={x:xData,y:yData}))
    print("y = %fx +%f" % (s.run(a),s.run(b))) # x는 10을향해 가야함, y는 0을향해 가야함

bestA = s.run(a)[0]
bestB = s.run(b)[0]
print(bestA,bestB)
```

```python
# 4) 완성된 알고리즘 써서 미지의 데이터에 대한 결과 예측하기위한 작업

newX = float(input("x : "))
newX =[newX]
result = s.run(sik, feed_dict ={x:newX})[0]
print(result) # 4를 넣었을때 나오는 예측값
```
