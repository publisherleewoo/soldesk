```python
import tensorflow as tf

tf.compat.v1.disable_v2_behavior() # tf 2.x기능 off    1버전대 기능 먼저 공부

```

```python
# 1) 인공신경망 구성
a = tf.constant(10) #상수
b = tf.constant(20) #상수
c = tf.add(a,b)
d = tf.constant("ㅋ")

print(a,c,d) # 실제로 10이 들어감 x, 실제로 덧셈 x

# 2) 학습데이터를 넣어서 최적의 알고리즘 찾기
s = tf.compat.v1.Session() #세션실행
aResult = s.run(a)
print(aResult)

dResult = s.run(d)
print(dResult.decode())

cResult = s.run(c)# 연관된거 다 실행됨
print(cResult)



# 완성된 알고리즘 써서 미지의 데이터에 대한 결과 예측하기위함이 최종목표 (다음챕터 계속)


```
