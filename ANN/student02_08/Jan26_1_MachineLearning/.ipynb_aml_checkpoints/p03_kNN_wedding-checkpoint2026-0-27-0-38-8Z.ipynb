```python
import numpy as np
import pandas as pd

df = pd.read_csv('./Users/student02_08/Jan26_1_MachineLearning/datingTestSet.txt', sep="\t", names=['비행기','게임','아이스크림','인기도'])

feature = df[['비행기','게임','아이스크림']].to_numpy()
label=df['인기도'].to_numpy()

print(feature)
print(label)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import MinMaxScaler

mm = MinMaxScaler()
feature = mm.fit_transform(feature) # 최대 최소 비율로 표현하게 변경

knc = KNeighborsClassifier(10) # 가장가까운 top10

knc.fit(feature,label)
```

```python
airplane = float(input("비행기 :"))
game = float(input("게임 :"))
icecream = float(input("아이스크림 :"))

predictDate = np.array([[airplane,game,icecream]])

result = knc.predict(predictDate)[0]
print(result)
```
