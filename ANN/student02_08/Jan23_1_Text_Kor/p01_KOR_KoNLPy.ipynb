```python
sentence="컴퓨터 프로그램의 인터페이스를 테스트하기 위해 생성된 더미 텍스트입니다. 이 문장은 실제 의미를 담고 있지 않으며, 오직 텍스트가 화면에 어떻게 출력되는지 확인하기 위한 용도로 작성되었습니다. "
sentence+="폰트의 가독성이나 줄 간격, 그리고 레이아웃의 균형을 점검할 때 유용하게 활용할 수 있습니다. 데이터베이스에 샘플 데이터를 입력하거나 웹 페이지의 본문 영역을 채워 넣을 때 이 텍스트를 복사하여 사용하십시오. 프로그래밍 과정에서 텍스트의 길이가 충분한지 확인하는 작업은 매우 중요하며, 이를 통해 사용자 경험을 한층 더 개선할 수 있습니다. "
sentence+="계절이 지나가는 하늘에는 가을로 가득 차 있습니다. 나는 아무 걱정도 없이 가을 속의 별들을 다 헤아릴 듯합니다. "
sentence+="가슴 속에 하나둘 새겨지는 별을 이제 다 못 헤아리는 것은 쉬이 아침이 오는 까닭이요, "
sentence+="내일 밤이 남은 까닭이요, 아직 나의 청춘이 다하지 않은 까닭입니다.ㅎㅎㅎㅎ"

```

```python
from konlpy.tag import Okt

o = Okt() # Open Korea Text분석기 ( 구 트위터한글형태소분석기)

a = o.normalize(sentence) # 정규화 (정리)   ㅎㅎㅎㅎ가 ㅎㅎㅎ 세글자로 정리됨
print(a)

```

```python
b = o.phrases(sentence) # 어구 추출 (의미있는 단어로 나누기)
print(b)
```

```python
c = o.morphs(sentence) #형태소 분석(합성어 같은것도 나누기)
print(c)
```

```python
d = o.morphs(sentence,stem=True) #형태소 분석(합성어 같은것도 나누기 + 동사원형)
print(d)
```

```python
e = o.pos(sentence) #형태소 분석(합성어같읕것도 나누기) + 품사 태깅 : (단어, 품사)
print(e)

# for word,poomsa in e:
#     print(word,poomsa)
```

```python
f = o.pos(sentence,join=True) #형태소 분석(합성어같읕것도 나누기) + 품사 태깅 : 단어/품사
print(f)
# for wp in f:
#     print(wp)
```

```python
g = o.pos(sentence,stem=True) #형태소 분석(합성어같읕것도 나누기 + 동사원형) + 품사 태깅 : (단어, 품사)
print(g)
```
